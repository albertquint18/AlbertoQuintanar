```python
personas = int(input("Bienvenido al museo de A.H. ¿Cuántas personas son?:"))
for i in range(1, personas + 1):
    edad = int(input(f"Ingresa edad de la persona {i}: "))
    if edad >= 18:
        costo1 = 45
    elif edad >= 3:
        costo2 = 30
    else:
        costo3 = 0
    tipo = input(f"Ingresa el tipo de boleto (Adulto mayor = 4, Estudiante = 5, profesor = 6, otro = 7) para la persona {i}: ")
    while tipo == "4" and edad < 60:
        print("Ese tipo de boleto es para adultos mayores a 60 años, favor de ingresar un tipo valido")
        tipo = input(f"Ingresa el tipo de boleto (Adulto mayor = 4, Estudiante = 5, profesor = 6, otro = 7)")
        if tipo == "4" and edad >= 60:
                break
    if tipo == "4" and edad >= 60:
         precio_final = costo1 * 0.88
    elif tipo == "5" and edad < 18:
        precio_final = costo2 * 0.90
    elif tipo == "6" or (tipo == "5" and edad >= 18):
        precio_final = costo1 * 0.90
    elif tipo == "7" and edad >= 18:
        precio_final = costo1
    elif tipo == "7" and edad >= 3:
        precio_final = costo2
    else:
        precio_final = costo3
    print(f"El precio final del boleto para la persona {i} es: {precio_final}")
```
